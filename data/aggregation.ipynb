## Combine Monthly Datasets

Listings Dataset

```python
import pandas as pd

listings = []
data_counts = []

print("LISTINGS COUNTS ------------------------------------------------")

for i in range(4, 6):
    for j in range(1, 13):
        monthly = pd.read_csv(f"raw\\CRMLSListing202{i}{j:02d}.csv")
        listings.append(monthly)

        print(f"Read {len(monthly)} listings for 202{i}{j:02d}")
        data_counts.append(len(monthly))

for j in range(1, 9):
    monthly = pd.read_csv(f"raw\\CRMLSListing2026{j:02d}.csv", encoding="cp1252", low_memory=False)
    listings.append(monthly)

    print(f"Read {len(monthly)} listings for 2026{j:02d}")
    data_counts.append(len(monthly))

print("TOTAL LISTINGS COUNT -------------------------------------------")
print(sum(data_counts))

all_listings = pd.concat(listings, ignore_index=True)
print(f"Total count from concatenation: {len(all_listings)}")
all_listings.to_csv("processed\\CRMLSListing.csv", encoding="utf-8")

residential = all_listings[all_listings['PropertyType'] == 'Residential']
print(f"Total residential count: {len(residential)}")
residential.to_csv("processed\\CRMLSListing_Residential.csv", encoding="utf-8")
```
